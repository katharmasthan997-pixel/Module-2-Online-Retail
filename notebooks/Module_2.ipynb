```python
import numpy as np
```

```python
df=df[df['Quantity']>0]
df=df[df['UnitPrice']>0]
```

```python
df['UnitPrice'].unique()
```

```python
df['Total Price']=df['Quantity']*df['UnitPrice']
```

```python
df.columns
```

```python
df.head()
```

```python
df['Year']=df['InvoiceDate'].dt.year
df['Month']=df['InvoiceDate'].dt.month
df['Day']=df['InvoiceDate'].dt.day
df['Hour']=df['InvoiceDate'].dt.hour
```

```python
df.columns
```

```python
df['Customer Segment']=np.where(df['Total Price'] > 1000,'High','Low')
```

```python
df['Order Size']=np.where(df['Quantity']>10,'High','Low')
df['Day type']=np.where(df['Day'].isin([6,7]),'Weekend','Weekday')
```

```python
 """Analyze Quantity, UnitPrice, TotalPrice
● Calculate mean, median, and mode
● Find standard deviation, variance, and percentiles"""
```

```python
print("Quantity")
print(df['Quantity'].mean())
print(df['Quantity'].median())
print(df['Quantity'].mode())
print(df['Quantity'].std())
print(df['Quantity'].var())
print(df['Quantity'].quantile([0.25,0.5,0.75]))
```

```python
print("Unit Price")
print(df['UnitPrice'].mean())
print(df['UnitPrice'].median())
print(df['UnitPrice'].mode())
print(df['UnitPrice'].std())
print(df['UnitPrice'].var())
print(df['UnitPrice'].quantile([0.25,0.5,0.75]))
```

```python
print("Total Price")
print(df['Total Price'].mean())
print(df['Total Price'].median())
print(df['Total Price'].mode())
print(df['Total Price'].std())
print(df['Total Price'].var())
print(df['Total Price'].quantile([0.25,0.5,0.75]))
```

```python
import matplotlib.pyplot as plt

monthly_sales = df.groupby('Month')['Total Price'].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(range(1, 13))
plt.grid(True)

plt.show()
```

```python
country_sales = df.groupby('Country')['Total Price'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))
plt.bar(country_sales.index, country_sales.values)

plt.title('Top 10 Countries by Sales')
plt.xlabel('Country')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df['UnitPrice'], bins=50)

plt.title('Distribution of Unit Price')
plt.xlabel('Unit Price')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(df['Total Price'].dropna())

plt.title('Box Plot of Total Price')
plt.ylabel('Total Price')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Customer Segment')

plt.title('Customer Segment Distribution')
plt.xlabel('Customer Segment')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x='Customer Segment',
    y='Total Price'
)

plt.title('Total Price Distribution by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Price')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_cols = [
    'Quantity',
    'UnitPrice',
    'Total Price',
    'Year',
    'Month',
    'Day',
    'Hour'
]

correlation = df[numeric_cols].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Quantity',
    y='UnitPrice',
    hue='Customer Segment'
)

plt.title('Quantity vs Unit Price')
plt.xlabel('Quantity')
plt.ylabel('Unit Price')

plt.show()
```
